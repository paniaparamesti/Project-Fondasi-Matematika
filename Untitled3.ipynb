```python
print("=== SISTEM REGULASI CUTI KARYAWAN ===")

# Input persyaratan cuti
sisa_cuti = input("Apakah karyawan memiliki sisa cuti? (ya/tidak): ").lower()
persetujuan_atasan = input("Apakah atasan menyetujui cuti? (ya/tidak): ").lower()
kuota_cuti = input("Apakah kuota cuti departemen masih tersedia? (ya/tidak): ").lower()
masa_percobaan = input("Apakah karyawan sedang dalam masa percobaan? (ya/tidak): ").lower()

# Persetujuan HRD diperlukan jika karyawan dalam masa percobaan
if masa_percobaan == "ya":
    persetujuan_hrd = input("Apakah HRD menyetujui cuti? (ya/tidak): ").lower()
else:
    persetujuan_hrd = "ya"

# Aturan F1: syarat pengajuan dapat diproses
F1 = (sisa_cuti == "ya"
    and persetujuan_atasan == "ya"
    and kuota_cuti == "ya")

# Aturan F2: jika masa percobaan, harus mendapat persetujuan HRD
F2 = (masa_percobaan != "ya"
    or persetujuan_hrd == "ya")

# Keputusan akhir berdasarkan F1 dan F2
if F1 and F2:
    print("\nAjuan cuti anda disetujui")
else:
    print("\nMaaf, ajuan cuti anda belum dapat disetujui")
```
